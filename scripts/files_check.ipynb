```python

from pathlib import Path

ROOT = Path.cwd().resolve()

required_files = [
    "README.md",
    "requirements.txt",

    "01_problem_understanding.ipynb",
    "02_eda.ipynb",
    "03_data_cleaning.ipynb",
    "04_preprocessing.ipynb",
    "05_traditional_ml.ipynb",

    "src/data_processing.py",
    "src/traditional_models.py",
    "src/evaluation.py",

    "data/metadata/metadata_raw.csv",
    "data/metadata/metadata_inspected.csv",
    "data/metadata/metadata_clean.csv",
    "data/metadata/cleaning_log.csv",

    "data/splits/split_seed42.csv",
    "data/processed/preprocessing_config.json",

    "results/comparisons/model_results.csv",
]

required_directories = [
    "data/metadata",
    "data/splits",
    "data/processed",
    "models/traditional",
    "models/pretrained",
    "results/figures",
    "results/comparisons",
]

print("\n========== PROJECT FILE CHECK ==========\n")

missing_files = []
missing_dirs = []

for file in required_files:
    path = ROOT / file

    if path.exists():
        print(f"[OK]      {file}")
    else:
        print(f"[MISSING] {file}")
        missing_files.append(file)

print("\n========== DIRECTORY CHECK ==========\n")

for directory in required_directories:
    path = ROOT / directory

    if path.is_dir():
        print(f"[OK]      {directory}")
    else:
        print(f"[MISSING] {directory}")
        missing_dirs.append(directory)

print("\n========== PROCESSED ARRAY CHECK ==========\n")

arrays = [
    "data/processed/images_64.npy",
    "data/processed/images_128.npy",
    "data/processed/images_224.npy",
    "data/processed/labels.npy",
]

for array in arrays:
    path = ROOT / array

    if path.exists():
        print(f"[OK]      {array} ({path.stat().st_size / (1024**2):.2f} MB)")
    else:
        print(f"[MISSING] {array}")

print("\n========== SUMMARY ==========\n")
print(f"Missing files: {len(missing_files)}")
print(f"Missing directories: {len(missing_dirs)}")

if not missing_files and not missing_dirs:
    print("All required project files and directories are present.")
else:
    print("Some project components are missing. Review the list above.")
```
