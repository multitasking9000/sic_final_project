```python

from pathlib import Path
import zipfile

PROJECT_ROOT = Path.cwd().resolve()

ZIP_FILE = PROJECT_ROOT / "Cats and Dogs Dataset for Image Classification.zip"
DATASET_DIR = PROJECT_ROOT / "Cats and Dogs Classification Dataset"

if not ZIP_FILE.exists():
    raise FileNotFoundError(f"Dataset ZIP not found: {ZIP_FILE}")

if DATASET_DIR.exists() and any(DATASET_DIR.iterdir()):
    print(f"Dataset already extracted: {DATASET_DIR}")
else:
    print("Extracting dataset...")

    with zipfile.ZipFile(ZIP_FILE, "r") as zip_ref:
        zip_ref.extractall(PROJECT_ROOT)

    print("Extraction completed!")

for folder in ["Cat", "Dog"]:
    folder_path = DATASET_DIR / folder

    if not folder_path.exists():
        print(f"WARNING: Missing folder: {folder}")
    else:
        count = sum(1 for f in folder_path.iterdir() if f.is_file())
        print(f"{folder}: {count} files")

print("Dataset extraction check completed.")
```
